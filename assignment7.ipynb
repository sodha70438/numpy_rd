```python
import numpy as np

zomato = np.array([4.5, 3.8, 4.2, 4.7, 3.9])
swiggy = np.array([4.1, 4.6, 3.7, 4.3, 4.8])

ratings = np.concatenate((zomato, swiggy))

print(ratings)
```

```python
messages = np.array([
    [10, 20, 15, 25, 30, 20, 10],
    [5, 10, 8, 12, 15, 10, 5],
    [20, 25, 30, 20, 35, 40, 25]
])

new_user = [15, 20, 10, 25, 20, 15, 30]
messages = np.insert(messages, 3, new_user, axis=0)

print("After Insert:")
print(messages)

total_messages = np.sum(messages, axis=1)
least_user = np.argmin(total_messages)

messages = np.delete(messages, least_user, axis=0)

print("\nAfter Delete:")
print(messages)
```

```python
import numpy as np

views = np.array([1000, 2000, 3000, 4000, 5000])

view_data = views.view()
view_data[0] = 1500

print("Original after View:", views)
print("View:", view_data)

copy_data = views.copy()
copy_data[0] = 9999

print("Original after Copy:", views)
print("Copy:", copy_data)
```
