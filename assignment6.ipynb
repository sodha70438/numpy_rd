```python
import numpy as np

prices = np.array([199, 299, 399, 499, 599])

print("First Price:", prices[0])
print("Last Price:", prices[-1])
```

```python
scores = np.array([
    [50, 70, 80, 90, 60],
    [40, 60, 75, 85, 95],
    [30, 55, 65, 70, 80]
])

result = scores[:, 1:4]

print(result)
```

```python
ratings = np.array([4.5, 3.8, 4.2, 2.9, 5.0, 3.5])

last_three = ratings[-3:]

print(last_three)
```

```python
numbers = np.arange(1, 21)

result = numbers[1::3]

print(result)
```

```python
prices = np.array([299, 599, 450, 799, 999, 350, 650])

result = prices[prices > 500]

print(result)
```
