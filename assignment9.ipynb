```python
import numpy as np

ratings = np.array([4.5, 3.8, np.nan, 4.2, np.nan, 5.0, 3.5])

missing = np.isnan(ratings)

print("Missing Ratings:", missing)
print("Total Missing Ratings:", np.sum(missing))
```

```python
import numpy as np

prices = np.array([50, 150, 500, 1200, 750, 80, 1500])

result = np.clip(prices, 100, 1000)

print(result)
```

```python
import numpy as np

views = np.array([1000, np.nan, 5000, np.inf, 3000, 8000, np.nan])
max_value = np.max(views[np.isfinite(views)])

views[np.isnan(views)] = 0
views[np.isinf(views)] = max_value

print(views)
```
